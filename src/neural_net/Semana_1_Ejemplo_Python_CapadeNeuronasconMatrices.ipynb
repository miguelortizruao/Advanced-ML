**Procesamiento de un Neurona Artificial**

Vamos a mostrar como es el procesamiento de informacioón que tiene un neurona artificial realizando una implementación matricial usando Numpy

**Definición del vector de entrada X**

Inicialmente creamos el vector de entrada X

```
X=np.array([[1],[0.5],[-0.6]])
print(X)
```

**Definición del vector de conexiones W X**

Posteriormente creamos el vector de pesos o conexiones sinápticas W

```
W=np.array([[1.0],[-0.7],[-1.0]])
print(W)
```

**Cálculo de la entrada neta**

Con el vector de entrada y de pesos sinápticos procedemos a cálculo de la entrada neta que tiene la neurona

```
Neta=(np.sum(np.multiply(X,W)))
print(Neta)
```

**Cálculo de la salida de la neurona**

Dicha entrada neta se evalua en la Function de activacion para generar la salida, en este caso usaremos al función de activación tipo sigmoidal

```
Y=1.0/(1+np.exp(-Neta))
print(Y)
```
