```python
# Importación de librerías esenciales para cálculo, visualización y machine learning
import numpy as np  # Operaciones numéricas, matrices y vectores
import matplotlib.pyplot as plt  # Creación y despliegue de gráficos
import pandas as pd  # Manejo y estructuración de datos en tablas (DataFrames)

from sklearn.tree import DecisionTreeClassifier  # Árbol de decisión para clasificación
from sklearn.ensemble import AdaBoostClassifier  # Algoritmo de ensamble secuencial AdaBoost (Adaptive Boosting)
from sklearn.datasets import load_breast_cancer  # Dataset de cáncer de mama para evaluación
from sklearn.model_selection import train_test_split  # Función para dividir datos en entrenamiento y prueba
```

```python
# 1. Cargar el dataset de cáncer de mama (569 pacientes, 30 características clínicas)
cancer = load_breast_cancer()

# 2. Dividir los datos en conjunto de entrenamiento (75%) y conjunto de prueba (25%)
# - cancer.data: matriz de características (X)
# - cancer.target: diagnósticos reales (y: 0=maligno, 1=benigno)
# - random_state=0: fija la semilla para que la partición sea reproducible
X_train, X_test, y_train, y_test = train_test_split(
    cancer.data, cancer.target, random_state=0)
```

```python
# Definición del 'estimador débil' (Weak Learner) base para AdaBoost:
# - criterion='entropy': mide la ganancia de información usando entropía de Shannon en lugar de Gini
# - random_state=1: reproducibilidad
# - max_depth=1: árbol extremadamente simple de 1 solo nivel (llamado 'Decision Stump' o muñón)
#   (Solo hace una sola pregunta binaria para evitar que memorice datos por sí solo)
tree = DecisionTreeClassifier(criterion='entropy',
                              random_state=1,
                              max_depth=1)
```

```python
# Creación del ensamble AdaBoost (Adaptive Boosting):
# - base_estimator=tree: el modelo base débil que se entrenará repetidamente (árbol con profundidad 1)
# - n_estimators=500: entrenará una secuencia de 500 árboles débiles, cada uno corrigiendo los errores del anterior
# - learning_rate=0.1: tasa de aprendizaje (controla qué tan agresivamente se ajustan los pesos en cada paso)
# - random_state=1: fija la semilla para reproducibilidad
ada = AdaBoostClassifier(base_estimator=tree,
                         n_estimators=500,
                         learning_rate=0.1,
                         random_state=1)

# Entrenar el ensamble de forma secuencial con los datos de entrenamiento
ada.fit(X_train, y_train)

# Evaluar la exactitud (accuracy) en entrenamiento y en prueba:
# Compara cómo subió el rendimiento frente a un solo árbol (test sube a 96.5%):
print("Accuracy on training set: {:.3f}".format(ada.score(X_train, y_train)))
print("Accuracy on test set: {:.3f}".format(ada.score(X_test, y_test)))
```
